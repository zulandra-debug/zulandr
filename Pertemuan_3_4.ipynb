```python
#Tugas soal 1

jumlah_buku = int(input("Masukkan jumlah buku: "))
harga_buku = int(input("Masukkan harga per buku: Rp "))
persen_diskon = float(input("Masukkan diskon (%): "))
uang_budi = int(input("Masukkan uang Budi: Rp "))
total = jumlah_buku * harga_buku
diskon = int(total * persen_diskon / 100)
bayar = total - diskon
kembalian = uang_budi - bayar
per_orang = bayar // 4
sisa = bayar % 4
uang_cukup = uang_budi >= bayar

print("\n=== HASIL PERHITUNGAN ===")
print("Total       :", total)
print("Diskon      :", diskon)
print("Bayar       :", bayar)
print("Kembalian   :", kembalian)
print("Per orang   :", per_orang)
print("Sisa        :", sisa)
print("Uang cukup? :", uang_cukup)
```

```python
#tugas soal 2

jumlah_panel = int(input("Masukkan jumlah panel surya: "))
daya_panel = float(input("Masukkan daya setiap panel (Wp): "))
jam_penyinaran = float(input("Masukkan jam penyinaran per hari: "))
efisiensi = float(input("Masukkan efisiensi sistem (%): "))
kebutuhan_harian = float(input("Masukkan kebutuhan listrik per hari (kWh): "))
tarif_listrik = float(input("Masukkan tarif listrik per kWh (Rp): "))
total_daya = (jumlah_panel * daya_panel) / 1000
energi_harian = total_daya * jam_penyinaran * (efisiensi / 100)
energi_bulanan = energi_harian * 30
nilai_listrik = energi_bulanan * tarif_listrik
kekurangan = max(0, kebutuhan_harian - energi_harian)
mencukupi = energi_harian >= kebutuhan_harian

print("\n=== HASIL PERHITUNGAN PLTS ===")
print("Energi harian (kWh) :", round(energi_harian, 2))
print("Energi bulanan (kWh):", round(energi_bulanan, 2))
print("Nilai listrik (Rp)  :", round(nilai_listrik, 2))
print("Kekurangan (kWh)    :", round(kekurangan, 2))
print("Mencukupi?          :", mencukupi)
```

```python
#Tugas soal 3

daya_generator = float(input("Masukkan daya generator (kW): "))
jam_operasi = float(input("Masukkan jam operasi per hari: "))
konsumsi_solar = float(input("Masukkan konsumsi solar (liter/kWh): "))
harga_solar = float(input("Masukkan harga solar per liter (Rp): "))
kapasitas_tangki = float(input("Masukkan kapasitas tangki (liter): "))
arus = float(input("Masukkan arus kabel (A): "))
hambatan = float(input("Masukkan hambatan kabel (ohm): "))
energi = daya_generator * jam_operasi
solar = energi * konsumsi_solar
biaya = solar * harga_solar
tangki_cukup = kapasitas_tangki // solar
sisa_solar = kapasitas_tangki % solar
rugi_kabel = arus ** 2 * hambatan
rugi_tinggi = (rugi_kabel > 150) and (arus >= 20)

print("\n=== HASIL PERHITUNGAN ===")
print("Energi (kWh)       :", round(energi, 2))
print("Solar (liter)      :", round(solar, 2))
print("Biaya (Rp)         :", round(biaya, 2))
print("Tangki cukup       :", int(tangki_cukup), "hari")
print("Sisa solar (L)     :", round(sisa_solar, 2))
print("Rugi kabel (W)     :", round(rugi_kabel, 2))
print("Rugi tinggi?       :", rugi_tinggi)
```
